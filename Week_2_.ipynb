## Task 8:Merge, Concatenate,GroupBy & Pivot Table

```python
#Merge two data frames using a common column (Customer Id)
import pandas as pd
#first dataset
customer=pd.DataFrame({'customer_Id':[1,2,3,],
                       'Name':['Alice','Bob','Charlie']})
order=pd.DataFrame({
    'customer_Id':[1,2,3],
    'Amount':[250,400,350]})

merged_inner=pd.merge(customer,order,on='customer_Id',how='inner')
print(merged_inner)
```

```python
#Concatinate 2 dataframes
jan_sales=pd.DataFrame({'items':["Laptop","Phone"],'sales':[10,20]})
feb_sales=pd.DataFrame({'items':["Keyboard","Monitor"],'sales':[10,20]})
all_sales=pd.concat([jan_sales,feb_sales])
print(all_sales)
```

```python
#Group by
total_amount=order.groupby('customer_Id')['Amount'].sum()
print(total_amount)
```

```python
#Calculate aggregate values such as sum,mean,count,minimumm or maximum
jan_sales=pd.DataFrame({'items':["Laptop","Phone"],'sales':[10,20]})
feb_sales=pd.DataFrame({'items':["Keyboard","Monitor"],'sales':[10,20]})
all_sales=pd.concat([jan_sales,feb_sales])
print(all_sales)

print("\nsum of sales:",all_sales['sales'].sum())
print("mean of sales:",all_sales['sales'].mean())
print("count of sales:",all_sales['sales'].count())
print("Maximum of sales:",all_sales['sales'].max())


```

```python
#Pivot Table
jan_sales=pd.DataFrame({'items':["Laptop","Phone"],'sales':[10,20],'month':"Jan"})
feb_sales=pd.DataFrame({'items':["Keyboard","Monitor"],'sales':[10,20],'month':"Feb"})
all_sales=pd.concat([jan_sales,feb_sales])
print(all_sales)
print("\n")
#Creating the pivot Table
print("Pivot table using month:")
pivot=pd.pivot_table(
    all_sales,values='sales',
    index='items',
    columns='month',
    aggfunc='sum',
    fill_value=0)
print(pivot)

```

```python
#Summarize data based on rows , columns and values
summary_pivot = pd.pivot_table(
    all_sales,
    index="items",
    values="sales",
    aggfunc=["sum", "mean", "count", "min", "max"]
)

print(summary_pivot)
```

```python
import numpy as np
import pandas as pd

# ==========================================
# 1. LOADING THE DATASET
# ==========================================
data = {
    "Order_ID": np.arange(101, 109),
    "Date": ["2026-01-05", "2026-01-12", "2026-01-18", "2026-02-02",
             "2026-02-10", "2026-02-15", "2026-02-20", "2026-02-25"],
    "Region": ["North", "South", "East", "North", "West", np.nan, "East", "South"],
    "Category": ["Electronics", "Furniture", "Electronics", "Office",
                 "Furniture", "Electronics", "Office", "Electronics"],
    "Product": ["Laptop", "Desk Chair", "Phone", "Printer",
                "Bookshelf", "Monitor", "Desk Lamp", "Tablet"],
    "Units_Sold": [5, 12, np.nan, 4, 7, 10, 15, 8],
    "Unit_Price": [800, 150, 600, 200, 120, 250, np.nan, 400]
}
df = pd.DataFrame(data)

# ==========================================
# 2. DATA INSPECTION
# ==========================================
print("--- First 5 Rows ---")
print(df.head())

print("\n--- Dataset Info & Data Types ---")
print(df.info())

print("\n--- Summary Statistics ---")
print(df.describe())

# ==========================================
# 3. IDENTIFYING & HANDLING MISSING VALUES
# ==========================================
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

df["Region"] = df["Region"].fillna("Unknown")
df["Units_Sold"] = df["Units_Sold"].fillna(np.nanmedian(df["Units_Sold"]))
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].mean())

df["Total_Revenue"] = np.multiply(df["Units_Sold"], df["Unit_Price"])

# ==========================================
# 4. SELECTING AND FILTERING DATA
# ==========================================
product_revenue = df.loc[:, ("Product", "Category", "Total_Revenue")]

high_value_orders = df[df["Total_Revenue"] > 1500]
print("\n--- High Value Orders (> $1,500) ---")
print(high_value_orders.loc[:, ("Order_ID", "Product", "Total_Revenue")])

# ==========================================
# 5. SORTING DATA
# ==========================================
sorted_df = df.sort_values(by="Total_Revenue", ascending=False)
print("\n--- Top Orders by Revenue ---")
print(sorted_df.loc[:, ("Product", "Region", "Total_Revenue")].head(3))

# ==========================================
# 6. GROUPBY ANALYSIS
# ==========================================
category_summary = df.groupby("Category").agg(
    Total_Units=("Units_Sold", "sum"),
    Total_Revenue=("Total_Revenue", "sum"),
    Avg_Order_Value=("Total_Revenue", "mean")
).reset_index()

print("\n--- Sales by Category (GroupBy) ---")
print(category_summary)

# ==========================================
# 7. PIVOT TABLE
# ==========================================
pivot_sales = pd.pivot_table(
    df,
    values="Total_Revenue",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)
print("\n--- Revenue Pivot Table (Region vs. Category) ---")
print(pivot_sales)

# ==========================================
# 8. USEFUL INSIGHTS
# ==========================================
top_category = category_summary.loc[category_summary["Total_Revenue"].idxmax(), "Category"]
top_product = sorted_df["Product"].values.item(0)
overall_revenue = np.sum(df["Total_Revenue"])

print("\n--- Key Insights ---")
print(f"1. Overall Revenue: ${overall_revenue:,.2f}")
print(f"2. Top Performing Category: {top_category}")
print(f"3. Highest Revenue Product: {top_product}")

# ==========================================
# 9. EXPORTING THE CLEANED DATASET
# ==========================================
df.to_csv("cleaned_sales_data.csv", index=False)
print("\nCleaned dataset exported to 'cleaned_sales_data.csv'.")
```
